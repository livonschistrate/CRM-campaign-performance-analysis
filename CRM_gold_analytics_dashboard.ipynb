
### 📈 Aggregation Layer (for Dashboard Performance)

**`agg_campaign_daily`**
- **Purpose**: Pre-aggregated metrics for fast dashboard queries
- **Grain**: Daily metrics by product, category, channel, campaign
- **Metrics**: 
  - Volume: `customers_reached`, `communications_sent`, `delivered_count`, `opened_count`, `clicked_count`, `sales_count`
  - Performance: `delivery_rate`, `open_rate`, `click_rate`, `conversion_rate`, `revenue`
  - Timing: `avg_days_to_sale`
- **Update Pattern**: Full refresh daily (can be incremental in production)

````python
# Create Aggregated Reporting Table for Dashboard
# Pre-aggregate metrics by date, product, channel for fast dashboard queries

from pyspark.sql.functions import sum as spark_sum, count as spark_count, countDistinct, avg, round as spark_round, when, col

agg_campaign_daily = (
    fact_crm_communication
    .join(dim_date, 'date_key', 'left')
    .join(df_products, fact_crm_communication.product_key == df_products.product_id, 'left')
    .join(df_campaigns, fact_crm_communication.campaign_key == df_campaigns.campaign_id, 'left')
    .join(dim_channel, fact_crm_communication.channel_key == dim_channel.channel_id, 'left')
    .groupBy(
        'date_key',
        dim_date.date.alias('date'),
        dim_date.year,
        dim_date.month,
        dim_date.week,
        dim_date.month_name,
        df_products.product_name,
        df_products.category,
        dim_channel.channel_name.alias('channel'),
        fact_crm_communication.channel_key,
        fact_crm_communication.campaign_key
    )
    .agg(
        countDistinct('customer_key').alias('customers_reached'),
        spark_count('communication_id').alias('communications_sent'),
        spark_sum('delivered_count').alias('delivered_count'),
        spark_sum('opened_count').alias('opened_count'),
        spark_sum('clicked_count').alias('clicked_count'),
        spark_sum('converted_count').alias('sales_count'),
        spark_sum('communication_cost').alias('campaign_cost'),
        spark_sum('sale_amount').alias('revenue'),
        avg('days_to_sale').alias('avg_days_to_sale')
    )
    .withColumn('delivery_rate',
                when(
                    col('communications_sent') > 0,
                    spark_round(col('delivered_count') / col('communications_sent') * 100, 2)
                    )
                )
    .withColumn('open_rate', 
                when(
                    col('delivered_count') > 0, 
                    spark_round(col('opened_count') / col('delivered_count') * 100, 2)
                    )
                )
    .withColumn('click_rate',
                when(
                    col('opened_count') > 0,
                    spark_round(col('clicked_count') / col('opened_count') * 100, 2)
                    )
                )
    .withColumn('conversion_rate',
                when(
                    col('communications_sent') > 0,
                    spark_round(col('sales_count') / col('communications_sent') * 100, 2)
                    )
                )
    .withColumn('roi', 
                when(
                    col('campaign_cost') > 0,
                    spark_round(((col('revenue') - col('campaign_cost')) / col('campaign_cost')) * 100, 2)
                    ).otherwise(None)
                )
    .orderBy('date', 'product_name', 'channel')
)

print(f"\nAggregated Reporting Table: {agg_campaign_daily.count()} records")
print(f"\nGrain: Daily aggregations by product, category, channel, campaign")
print(f"\nSample Dashboard Metrics:")
display(
    agg_campaign_daily
    .select('date', 'product_name', 'channel', 'customers_reached', 
            'communications_sent', 'sales_count', 'revenue', 'conversion_rate')
    .orderBy('revenue', ascending=False)
    .limit(10)
)
````

````python
%sql
WITH funnel_metrics AS (
    SELECT 
        COUNT(f.communication_id) as total_sent,
        SUM(f.delivered_count) as total_delivered,
        SUM(f.opened_count) as total_opened,
        SUM(f.clicked_count) as total_clicked,
        SUM(f.converted_count) as total_converted,
        AVG(CASE WHEN f.converted_count > 0 THEN f.days_to_sale END) as avg_days_to_conversion
    FROM 
        workspace.crm_info.fact_crm_communication f
)
SELECT 
    'Sent' as stage,
    total_sent as count,
    100.0 as percent_of_sent,
    NULL as conversion_rate
FROM funnel_metrics

UNION ALL

SELECT 
    'Delivered' as stage,
    total_delivered as count,
    total_delivered / total_sent * 100.0 as percent_of_sent,
    NULL as conversion_rate
FROM funnel_metrics

UNION ALL

SELECT 
    'Opened' as stage,
    total_opened as count,
    total_opened / total_sent * 100.0 as percent_of_sent,
    total_opened / NULLIF(total_delivered, 0) as conversion_rate
FROM funnel_metrics

UNION ALL

SELECT 
    'Clicked' as stage,
    total_clicked as count,
    total_clicked / total_sent * 100.0 as percent_of_sent,
    total_clicked / NULLIF(total_opened, 0) as conversion_rate
FROM funnel_metrics

UNION ALL

SELECT 
    'Converted' as stage,
    total_converted as count,
    total_converted / total_sent * 100.0 as percent_of_sent,
    total_converted / NULLIF(total_clicked, 0) as conversion_rate
FROM funnel_metrics

ORDER BY 
    CASE stage 
        WHEN 'Sent' THEN 1
        WHEN 'Delivered' THEN 2
        WHEN 'Opened' THEN 3
        WHEN 'Clicked' THEN 4
        WHEN 'Converted' THEN 5
    END;
````

````python
%sql
SELECT
  dc.campaign_name,
  dch.channel_name,
  SUM(f.communication_cost) AS total_cost,
  SUM(f.converted_count) AS conversions,
  SUM(f.sale_amount) AS revenue,
  ROUND(SUM(f.communication_cost) / NULLIF(SUM(f.converted_count), 0), 2) AS cost_per_conversion,
  ROUND(SUM(f.sale_amount) / NULLIF(SUM(f.converted_count), 0), 2) AS revenue_per_conversion,
  ROUND(
    (SUM(f.sale_amount) - SUM(f.communication_cost)) / NULLIF(SUM(f.communication_cost), 0),
    2
  ) AS roi_ratio,
  CASE
    WHEN
      ROUND(SUM(f.communication_cost) / NULLIF(SUM(f.converted_count), 0), 2) < 1.0
      AND ROUND(SUM(f.sale_amount) / NULLIF(SUM(f.converted_count), 0), 2) > 20000
    THEN
      'High Efficiency'
    WHEN
      ROUND(SUM(f.communication_cost) / NULLIF(SUM(f.converted_count), 0), 2) > 5.0
      AND ROUND(SUM(f.sale_amount) / NULLIF(SUM(f.converted_count), 0), 2) < 20000
    THEN
      'Low Efficiency'
    ELSE 'Medium Efficiency'
  END AS efficiency_category
FROM
  workspace.crm_info.fact_crm_communication f
    JOIN workspace.crm_info.dim_campaign dc
      ON f.campaign_key = dc.campaign_id
    JOIN workspace.crm_info.dim_channel dch
      ON f.channel_key = dch.channel_id
GROUP BY
  dc.campaign_name,
  dch.channel_name
HAVING
  conversions > 10
````

````python
%sql
WITH customer_ltv_segments AS (
  SELECT
    dc.customer_id,
    dc.segment,
    dc.region,
    dc.lifetime_value,
    CASE
      WHEN dc.lifetime_value >= 75000 THEN 'High LTV (75K+)'
      WHEN dc.lifetime_value >= 50000 THEN 'Medium LTV (50-75K)'
      WHEN dc.lifetime_value >= 25000 THEN 'Low LTV (25-50K)'
      ELSE 'Very Low LTV (<25K)'
    END as ltv_segment
  FROM
    workspace.crm_info.dim_customer dc
)
SELECT
  cls.ltv_segment,
  cls.segment as customer_segment,
  COUNT(DISTINCT f.customer_key) as customers_reached,
  COUNT(f.communication_id) as total_communications,
  ROUND(
    COUNT(f.communication_id) * 1.0 / COUNT(DISTINCT f.customer_key),
    1
  ) as avg_comms_per_customer,
  SUM(f.converted_count) as conversions,
  SUM(f.converted_count) / COUNT(f.communication_id) as conversion_rate,
  SUM(f.sale_amount) as total_revenue,
  ROUND(AVG(cls.lifetime_value), 2) as avg_ltv,
  ROUND(SUM(f.sale_amount) / NULLIF(COUNT(DISTINCT f.customer_key), 0), 2) as revenue_per_customer
FROM
  workspace.crm_info.fact_crm_communication f
    JOIN customer_ltv_segments cls
      ON f.customer_key = cls.customer_id
GROUP BY
  cls.ltv_segment,
  cls.segment
ORDER BY
  avg_ltv DESC
````

````python
%sql
WITH first_contact AS (
  SELECT
    customer_key,
    MIN(event_date) as first_contact_date,
    DATE_TRUNC('MONTH', MIN(event_date)) as cohort_month
  FROM
    workspace.crm_info.fact_crm_communication
  GROUP BY
    customer_key
),
cohort_activity AS (
  SELECT
    fc.cohort_month,
    DATE_TRUNC('MONTH', f.event_date) as activity_month,
    CAST(
      FLOOR(MONTHS_BETWEEN(f.event_date, fc.first_contact_date)) AS INT
    ) as months_since_first_contact,
    COUNT(DISTINCT f.customer_key) as active_customers,
    SUM(f.converted_count) as conversions,
    SUM(f.sale_amount) as revenue
  FROM
    workspace.crm_info.fact_crm_communication f
      JOIN first_contact fc
        ON f.customer_key = fc.customer_key
  GROUP BY
    fc.cohort_month,
    DATE_TRUNC('MONTH', f.event_date),
    CAST(FLOOR(MONTHS_BETWEEN(f.event_date, fc.first_contact_date)) AS INT)
),
cohort_size AS (
  SELECT
    cohort_month,
    COUNT(DISTINCT customer_key) as cohort_size
  FROM
    first_contact
  GROUP BY
    cohort_month
)
SELECT
  ca.cohort_month,
  ca.activity_month,
  ca.months_since_first_contact,
  cs.cohort_size,
  ca.active_customers,
  ca.active_customers / cs.cohort_size as retention_rate,
  ca.conversions,
  ROUND(ca.revenue, 2) as revenue
FROM
  cohort_activity ca
    JOIN cohort_size cs
      ON ca.cohort_month = cs.cohort_month
ORDER BY
  ca.cohort_month,
  ca.months_since_first_contact
````

````python
%sql
WITH historical_revenue AS (
  SELECT
    dd.date as forecast_date,
    SUM(f.sale_amount) as revenue,
    NULL as revenue_lower,
    NULL as revenue_upper,
    'Historical' as data_type
  FROM
    workspace.crm_info.fact_crm_communication f
      JOIN workspace.crm_info.dim_date dd
        ON f.date_key = dd.date_key
  WHERE
    dd.date >= CURRENT_DATE - INTERVAL 90 DAYS
  GROUP BY
    dd.date
),
forecast_data AS (
  SELECT
    CAST(forecast_date AS DATE) as forecast_date,
    predicted_revenue as revenue,
    revenue_lower,
    revenue_upper,
    'Forecast' as data_type
  FROM
    workspace.crm_info.forecast_campaign_performance
  WHERE
    forecast_date >= CURRENT_DATE
),
last_historical AS (
  SELECT
    forecast_date,
    revenue,
    NULL as revenue_lower,
    NULL as revenue_upper,
    'Forecast' as data_type
  FROM
    historical_revenue
  ORDER BY
    forecast_date DESC
  LIMIT 1
)
SELECT
  *
FROM
  historical_revenue
UNION ALL
SELECT
  *
FROM
  last_historical
UNION ALL
SELECT
  *
FROM
  forecast_data
ORDER BY
  forecast_date
````

````python
%sql
SELECT 
    dp.category,
    dch.channel_name,
    COUNT(DISTINCT f.customer_key) as customers_reached,
    SUM(f.converted_count) as conversions,
    SUM(f.sale_amount) as revenue,
    SUM(f.converted_count) / COUNT(f.communication_id) as conversion_rate
FROM 
    workspace.crm_info.fact_crm_communication f
JOIN 
    workspace.crm_info.dim_campaign dc ON f.campaign_key = dc.campaign_id
JOIN 
    workspace.crm_info.dim_product dp ON f.product_key = dp.product_id
JOIN 
    workspace.crm_info.dim_channel dch ON f.channel_key = dch.channel_id
GROUP BY 
    dp.category, dch.channel_name
ORDER BY 
    revenue DESC
````

## 🥇 Gold Layer - Aggregation & Analytics Documentation

### Purpose

The **Gold Layer** provides **pre-aggregated, dashboard-optimized tables** and **advanced analytics** for business intelligence consumption. This layer sits on top of the Silver star schema and focuses on:
1. **Performance optimization** via pre-computed aggregations
2. **Advanced analytics** (cohorts, LTV, fatigue analysis)
3. **Business-ready datasets** for dashboard widgets

---

### 📊 Tables Created

#### **Aggregation Tables** (Pre-computed for Dashboard Performance)

**`agg_campaign_daily`** - Daily pre-aggregated metrics
* **Grain**: One row per (date, product, category, channel, campaign) combination
* **Purpose**: Fast dashboard queries without hitting the fact table
* **Metrics**: 
  - Volume: customers_reached, communications_sent, delivered_count, opened_count, clicked_count, sales_count
  - Performance: delivery_rate, open_rate, click_rate, conversion_rate, revenue
  - Timing: avg_days_to_sale
* **Update Pattern**: Daily full refresh (or incremental in production)
* **Query Speed**: 100x faster than querying fact table directly

#### **Advanced Analytics Tables** (Dashboard-Specific)

These tables power the **Advanced Analytics** page of the dashboard:

* **Cohort Analysis** - Customer retention by signup month
* **LTV Impact** - Campaign performance segmented by customer lifetime value
* **Fatigue Analysis** - Conversion rates vs. communication frequency
* **Moving Averages** - 7-day smoothed trends for noise reduction
* **Segment Opportunity Matrix** - Segment performance (conversion rate vs revenue)
* **Campaign Efficiency Frontier** - ROI vs conversion rate scatter analysis

---

### 🎯 Key Design Decisions

1. **Pre-aggregation Strategy**: Dashboard widgets query `agg_campaign_daily` instead of 5,600-row fact table
   - **Benefit**: Sub-second dashboard load times
   - **Trade-off**: Daily refresh required to incorporate new data

2. **Denormalization for Speed**: Aggregation table includes denormalized attributes (product_name, category, channel, campaign_name)
   - **Benefit**: No joins needed in dashboard queries
   - **Trade-off**: Slightly larger storage footprint

3. **Grain Selection**: Daily grain balances detail vs. performance
   - **Too detailed** (hourly): Too many rows, slow queries
   - **Too coarse** (weekly): Loses time-based insights
   - **Just right** (daily): 61 unique dates × ~17 combinations = ~1,000 rows

4. **Computed Metrics**: Pre-calculate rates (delivery_rate, conversion_rate) to avoid division in every dashboard query

5. **Z-ORDER Optimization**: Fact and aggregation tables use Z-ORDER BY for clustered storage on frequently-filtered columns

---

### 📊 Typical Dashboard Query Patterns

**Pattern 1: Time-Series Trend** (uses agg_campaign_daily)
```sql
SELECT date, channel, communications_sent, sales_count, conversion_rate
FROM workspace.crm_info.agg_campaign_daily
WHERE date >= CURRENT_DATE - INTERVAL 30 DAYS
ORDER BY date, channel;
```

**Pattern 2: Top-N Ranking** (uses agg_campaign_daily)
```sql
SELECT product_name, 
       SUM(revenue) as total_revenue, 
       AVG(conversion_rate) as avg_conversion_rate
FROM workspace.crm_info.agg_campaign_daily

ORDER BY total_revenue DESC
LIMIT 10;
```

**Pattern 3: Weekly Roll-up** (uses agg_campaign_daily)
```sql
SELECT year, week, category, 
       SUM(communications_sent) as comms,
       SUM(sales_count) as sales
FROM workspace.crm_info.agg_campaign_daily
GROUP BY year, week, category
ORDER BY year, week;
```

**Pattern 4: ROI Analysis** (drill back to fact table for precise cost calculation)
```sql
SELECT dc.campaign_name,
       SUM(f.sale_amount) as revenue,
       SUM(f.communication_cost) as cost,
       (SUM(f.sale_amount) / NULLIF(SUM(f.communication_cost), 0)) as roi
FROM workspace.crm_info.fact_crm_communication f
JOIN workspace.crm_info.dim_campaign dc ON f.campaign_key = dc.campaign_id
GROUP BY dc.campaign_name;
```
